```python
import torch
import torch.nn as nn

from torch import distributions
```

Discrete

```python
test_logits = torch.randn(4, 4)
test_dist = distributions.Categorical(logits=test_logits)
test_dist
```

```python
def test():
    sample0, sample1 = [test_dist.sample() for _ in range(2)]
    log_prob0, log_prob1 = [test_dist.log_prob(s) for s in (sample0, sample1)]
    log_prob2 = test_dist.log_prob(sample0)
    return sample0, sample1, log_prob0, log_prob1, log_prob2

test()
```

```python
test_mean, _test_std = torch.randn(4, 8).chunk(2, -1)
test_std = torch.abs(_test_std)
test_dist = distributions.Normal(test_mean, test_std)
test_dist, test()
```

```python
test_mean, _test_std = torch.randn(4, 8).chunk(2, -1)
test_std = torch.abs(_test_std).diag_embed()
test_dist = distributions.MultivariateNormal(test_mean, scale_tril=test_std)
test_dist, test()
```
