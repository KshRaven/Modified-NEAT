```python
import numpy as np

from numba import njit, types, objmode
from numba.experimental import jitclass
from collections import OrderedDict
```

```python
FLOAT = types.float64
INT = types.int64
STR = types.unicode_type
BOOL = types.bool

Position = tuple[int, int]

POSITION = types.Tuple([INT, INT])
```

```python
@jitclass({
    'position': POSITION,
    'value': INT,
})
class Cell(object):
    def __init__(self, position: Position, value: int):
        self.position = position
        self.value    = value

    def __str__(self):
        with objmode(pos_str="unicode_type"):
            pos_str = str(self.position)
        return f"Cell({pos_str}, {self.value})"


test_cell = Cell((0, 0), 0)
```

```python
@njit
def use_cell():
    cell = Cell((1, 1), 3)
    print(f"Cell = {cell}")

use_cell()
```

```python
@njit
def use_cell_in_arr():
    cells = [Cell((i, j), (i*3) + (j+1)) for i in range(3) for j in range(3)]
    print([str(c) for c in cells])
    arr = np.array(cells, Cell)

use_cell_in_arr()
```
