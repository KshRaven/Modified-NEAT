```python
from numba import njit, types
from numba.extending import overload, register_jitable

@register_jitable
def clip(value, min, max):
    return np.clip(value, min, max)

@overload(clip)
def clip_numba(value, min, max):
    if isinstance(value, (types.Integer, types.Float)):
        return np.clip(value, min, max)
    elif isinstance(value, types.Array):
        return np.clip(value, min, max)
    else:
        raise ValueError('Unsupported value type')
```

```python
@njit
def run():
    return clip(np.random.randn(10, 3), -1, 1)
@njit
def run2():
    return clip(0, -1, 1)
```
