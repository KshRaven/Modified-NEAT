```python
players, length, width = 3, 3, 3
mask = np.random.rand(players, length, width) > 0.5  # random True/False
mask.astype(int)
```

```python
coords = []

for p in range(players):
    # Get indices where mask is True for this player
    ys, xs = np.where(mask[p])

    if len(xs) == 0:
        # no available positions
        coords.append(None)
    else:
        # pick a random available index
        i = np.random.randint(len(xs))
        coords.append((ys[i].item(), xs[i].item()))
coords
```

```python
fill = np.full_like(mask, False, bool)
for i, pos, in enumerate(coords):
    if pos is None:
        continue
    else:
        x, y = pos
        fill[i, x, y] = True
fill.astype(int)
```
